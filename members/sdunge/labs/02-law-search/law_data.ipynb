```python
### 형사 판례 목록 긁기 테스트
import requests
import pandas as pd

OC = "sdunge"  #인증키

# 판례 목록조회 API
url = "https://www.law.go.kr/DRF/lawSearch.do"
params = {
    "OC": OC,
    "target": "prec",      # 판례
    "type": "JSON",
    "query": "사기",        # 검색 키워드 (형사 죄명)
    "display": 20,         # 한 페이지에 20건
    "page": 1
}

res = requests.get(url, params=params)
data = res.json()

# 결과 확인
prec_list = data["PrecSearch"]["prec"]
print(f"받아온 판례 수: {len(prec_list)}건")
print(f"전체 검색 건수: {data['PrecSearch']['totalCnt']}건")

# 표로 보기
df = pd.DataFrame(prec_list)
df.head()
```

### CSV 파일 변환

```python
import time

OC = "sdunge"
url = "https://www.law.go.kr/DRF/lawSearch.do"

all_precs = []
keyword = "사기"       
max_pages = 20          

for page in range(1, max_pages + 1):
    params = {
        "OC": OC, "target": "prec", "type": "JSON",
        "query": keyword, "display": 20, "page": page
    }
    res = requests.get(url, params=params)
    data = res.json()
    precs = data["PrecSearch"].get("prec", [])
    if not precs:          # 더 이상 없으면 중단
        break
    all_precs.extend(precs)
    print(f"{page}페이지 완료 — 누적 {len(all_precs)}건")
    time.sleep(0.3)        # API 부담 안 주게 잠깐 쉬기

# 표로 정리
df = pd.DataFrame(all_precs)
print(f"\n총 {len(df)}건 수집 완료")
df.head()
```

```python
# raw 폴더에 csv파일 저장
save_path = "prec_fraud.csv"
df.to_csv(save_path, index=False, encoding="utf-8-sig")

print(f"저장 완료: {save_path}")
print(f"저장된 판례 수: {len(df)}건")
```

### 본문 테스트

```python
from elasticsearch import Elasticsearch, helpers
import pandas as pd

# 연결
es = Elasticsearch("http://localhost:9200")
print("연결 확인:", es.info()["version"]["number"])

# CSV 읽기 (노트북이 raw 폴더에 있으니 파일명만)
df = pd.read_csv("prec_fraud.csv")
print(f"판례 {len(df)}건 로드")

# 인덱스 이름
INDEX = "prec_search"

# 기존 인덱스 있으면 삭제하고 새로 (재실행 대비)
if es.indices.exists(index=INDEX):
    es.indices.delete(index=INDEX)
es.indices.create(index=INDEX)

# 판례를 하나씩 색인 (사건명 위주)
actions = []
for i, row in df.iterrows():
    actions.append({
        "_index": INDEX,
        "_id": i,
        "_source": {
            "사건명": str(row.get("사건명", "")),
            "사건번호": str(row.get("사건번호", "")),
            "선고일자": str(row.get("선고일자", "")),
            "판례일련번호": str(row.get("판례일련번호", "")),
        }
    })

helpers.bulk(es, actions)
es.indices.refresh(index=INDEX)
print(f"색인 완료: {es.count(index=INDEX)['count']}건")
```

```python
def search(keyword, size=5):
    res = es.search(index="prec_search", query={"match": {"사건명": keyword}}, size=size)
    print(f"\n'{keyword}' → {res['hits']['total']['value']}건")
    for hit in res["hits"]["hits"]:
        print(f"  [{hit['_score']:.2f}] {hit['_source']['사건명'][:50]}")

search("사기")
search("기망")
search("속여")
```

```python
import time

OC = "sdunge"
search_url = "https://www.law.go.kr/DRF/lawSearch.do"
detail_url = "https://www.law.go.kr/DRF/lawService.do"

# 1. 형사 판례 목록 
all_precs = []
for page in range(1, 11):   # 10페이지 = 약 200건
    params = {
        "OC": OC, "target": "prec", "type": "JSON",
        "query": "사기", "display": 20, "page": page,
        "search": 2, "prncYd": "20200101~20241231"
    }
    r = requests.get(search_url, params=params)
    precs = r.json()["PrecSearch"].get("prec", [])
    if not precs:
        break
    all_precs.extend(precs)
    print(f"{page}페이지 — 누적 {len(all_precs)}건")
    time.sleep(0.3)

print(f"\n목록 {len(all_precs)}건 수집\n")

# 2. 각 판례의 본문 긁기 (형사만)
results = []
for i, p in enumerate(all_precs):
    sid = p["판례일련번호"]
    r = requests.get(detail_url, params={"OC":OC,"target":"prec","ID":sid,"type":"JSON"})
    d = r.json()

    # 본문이 정상이고 형사인 것만
    if "PrecService" in d and isinstance(d["PrecService"], dict):
        svc = d["PrecService"]
        if svc.get("사건종류명") == "형사" and svc.get("판례내용"):
            results.append({
                "판례일련번호": sid,
                "사건명": svc.get("사건명",""),
                "법원명": svc.get("법원명",""),
                "선고일자": svc.get("선고일자",""),
                "판시사항": svc.get("판시사항",""),
                "판결요지": svc.get("판결요지",""),
                "참조조문": svc.get("참조조문",""),
                "참조판례": svc.get("참조판례",""),
                "판례내용": svc.get("판례내용",""),
            })
    if (i+1) % 20 == 0:
        print(f"본문 {i+1}건 처리 — 형사 확보 {len(results)}건")
    time.sleep(0.3)

print(f"\n최종 형사 판례(본문 포함): {len(results)}건")

df_body = pd.DataFrame(results)
df_body.head()
```
